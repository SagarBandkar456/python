```python
import random
jackpot=random.randint(1,100)
guess=int(input("Guess the number"))
counter=1

while guess!=jackpot:
    if guess<jackpot:
        print("Guess higher no")
    else:
        print("Guess lower no")

    guess=int(input("Guess the number"))
    counter+=1

print("You find correct no")    
print("You tool", counter, "attempts") 
```
